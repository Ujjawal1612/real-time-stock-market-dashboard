# StockSense — Runnable Demo
This notebook demonstrates the analysis pipeline using the included sample market dataset. The Streamlit dashboard uses live/near-real-time Yahoo Finance data.

```python
import pandas as pd
import matplotlib.pyplot as plt
from app.market import calculate_indicators, latest_snapshot
df = pd.read_csv('../data/sample_stock_data.csv', parse_dates=['Date'])
df.columns = [c.lower() for c in df.columns]
df.head()
```

```python
df = calculate_indicators(df)
snapshot = latest_snapshot(df)
snapshot
```

```python
print(f"Latest Price : {snapshot['price']:.2f}")
print(f"Change       : {snapshot['change']:.2f} ({snapshot['change_pct']:.2f}%)")
print(f"Volume       : {snapshot['volume']:,}")
print(f"RSI (14)     : {df['rsi_14'].dropna().iloc[-1]:.2f}")
```

```python
plt.figure(figsize=(12,5))
plt.plot(df['date'], df['close'], label='Close')
plt.plot(df['date'], df['sma_20'], label='SMA 20')
plt.plot(df['date'], df['sma_50'], label='SMA 50')
plt.title('StockSense — Price & Moving Averages')
plt.legend(); plt.tight_layout(); plt.show()
```

```python
plt.figure(figsize=(12,4))
plt.plot(df['date'], df['rsi_14'], label='RSI 14')
plt.axhline(70, linestyle='--'); plt.axhline(30, linestyle='--')
plt.title('RSI (14)'); plt.legend(); plt.tight_layout(); plt.show()
```
